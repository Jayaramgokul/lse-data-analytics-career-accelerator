## B. Data Preprocessing & Transformation

### Filtering Out Tweets Data with Missing Hashtags

```python
# Filter out rows where the 'tweet_entities_hashtags' column has missing values (NaN)
# This ensures that only tweets with hashtags are included in the filtered DataFrame
tweets_fil = tweets[~tweets["tweet_entities_hashtags"].isna()]

# Print the shape of the filtered DataFrame 'tweets_fil' to check the number of rows and columns
# This helps verify how many tweets with hashtags remain after filtering
print(tweets_fil.shape)
```

### Preprocessing Tweets Data for Hashtag and Date Analysis

```python
# Import necessary libraries
import pandas as pd
import datetime

# Step 1: Split hashtags into a list
# Split the hashtags column into lists (assuming they are comma-separated)
tweets_fil['hashtags_list'] = tweets_fil['tweet_entities_hashtags'].str.split(',')

# Step 2: Expand hashtags to individual rows
# Explode the hashtags list so each hashtag appears in its own row
tweets_exploded = tweets_fil.explode('hashtags_list')

# Step 3: Standardize hashtags
# Convert hashtags to lowercase for consistency
tweets_exploded['hashtags_list_lower'] = tweets_exploded['hashtags_list'].str.lower()
# Trim any leading or trailing spaces from the hashtags
tweets_exploded['hashtags_list_lower'] = tweets_exploded['hashtags_list_lower'].str.strip()

# Step 4: Define function to extract date from tweet_id
def get_date_from_tweet_id(tweet_id):
    # Ensure the tweet_id is an integer
    tweet_id = int(tweet_id)
    # Twitter epoch in milliseconds
    twitter_epoch = 1288834974657
    # Shift the tweet_id to remove lower 22 bits and calculate the timestamp in ms
    timestamp_ms = (tweet_id >> 22) + twitter_epoch
    # Convert milliseconds to seconds
    timestamp_s = timestamp_ms / 1000.0
    # Create a datetime object
    tweet_datetime = datetime.datetime.utcfromtimestamp(timestamp_s)
    # Return only the date part
    return tweet_datetime.date()

# Step 5: Apply date extraction function to get posting dates
# Add a new column 'Date' with the posting date derived from tweet_id
tweets_exploded["Date"] = tweets_exploded['tweet_id'].apply(get_date_from_tweet_id)
```

### Sentiment Analysis on Tweets Data

```python
import nltk
# Download the VADER lexicon for sentiment analysis (only needed once)
nltk.download('vader_lexicon')

from nltk.sentiment.vader import SentimentIntensityAnalyzer

# Initialize the VADER sentiment analyzer
sia = SentimentIntensityAnalyzer()

# Define a function to get sentiment scores for a given text using VADER
def get_sentiment(text):
    # Use VADER to analyze the text and return a dictionary of sentiment scores
    # Scores include 'neg' (negative), 'neu' (neutral), 'pos' (positive), and 'compound' (overall sentiment)
    sentiment = sia.polarity_scores(text)
    return sentiment

import re

# Define a function to preprocess the tweet text for better sentiment analysis
def preprocess_text(text):
    # Remove URLs by identifying patterns starting with 'http' or 'www'
    text = re.sub(r'http\S+|www.\S+', '', text)
    # Remove mentions (e.g., @username) and hashtags (e.g., #hashtag)
    text = re.sub(r'@\w+|#\w+', '', text)
    # Optionally remove special characters and numbers, retaining only letters and whitespace
    text = re.sub(r'[^A-Za-z\s]', '', text)
    # Strip leading and trailing whitespace for a cleaner result
    text = text.strip()
    return text

# Apply text preprocessing to each tweet, storing cleaned text in a new column 'clean_text'
tweets_exploded['clean_text'] = tweets_exploded['tweet_full_text'].apply(preprocess_text)

# Apply the sentiment analysis function to each cleaned tweet
# Store the sentiment scores (neg, neu, pos, and compound) in a new column 'sentiment_scores'
tweets_exploded['sentiment_scores'] = tweets_exploded['clean_text'].apply(get_sentiment)

# Extract the 'compound' score from the sentiment scores and store it in 'compound_score' column
# Compound score represents the overall sentiment ranging from -1 (most negative) to +1 (most positive)
tweets_exploded['compound_score'] = tweets_exploded['sentiment_scores'].apply(lambda x: x['compound'])

# Define a function to classify sentiment based on the compound score
def classify_sentiment(compound_score):
    # Classify as 'positive' if compound score >= 0.05
    if compound_score >= 0.05:
        return 'positive'
    # Classify as 'negative' if compound score <= -0.05
    elif compound_score <= -0.05:
        return 'negative'
    # Otherwise, classify as 'neutral'
    else:
        return 'neutral'

# Apply the classification function to each compound score
# Store the result in a new column 'sentiment' for easy sentiment analysis
tweets_exploded['sentiment'] = tweets_exploded['compound_score'].apply(classify_sentiment)
```

### 1.1.2 Appointment Status vs Appointment Mode



```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Step 1: Filter out 'Unknown' appointment modes before calculating 'attended_order'
filtered_data = ar[
    ((ar['appointment_status'] == 'Attended') | (ar['appointment_status'] == 'DNA')) &
    (ar['appointment_mode'] != 'Unknown')
]

# Step 2: Group by 'appointment_mode' and 'appointment_status' to get the count of appointments
grouped_data = filtered_data.groupby(['appointment_mode', 'appointment_status'], as_index=False)['count_of_appointments'].sum()

# Step 3: Calculate total appointments per 'appointment_mode' for percentage calculation
total_per_mode = grouped_data.groupby('appointment_mode')['count_of_appointments'].transform('sum')

# Step 4: Calculate the percentage of appointments for each 'appointment_status'
grouped_data['percentage'] = (grouped_data['count_of_appointments'] / total_per_mode) * 100

# Step 5: Pivot the data to prepare for stacked plotting
pivot_data = grouped_data.pivot(index='appointment_mode', columns='appointment_status', values=['count_of_appointments', 'percentage']).fillna(0)

# Define colors for the stacked bar plot
attended_color = '#4CAF50'  # Green for Attended
dna_color = '#F44336'       # Red for DNA

# Plot the stacked bar chart with percentage annotations
fig, ax = plt.subplots(figsize=(10, 6))

# Define x positions for bars
x = range(len(pivot_data))

# Plot 'Attended' counts
ax.bar(x, pivot_data['count_of_appointments']['Attended'], label='Attended', color=attended_color)

# Plot 'DNA' counts stacked on top of 'Attended'
ax.bar(x, pivot_data['count_of_appointments']['DNA'], bottom=pivot_data['count_of_appointments']['Attended'], label='DNA', color=dna_color)

# Set titles and labels
ax.set_title('Appointment Count and Percentage by Appointment Mode')
ax.set_xlabel('Appointment Mode')
ax.set_ylabel('Count of Appointments')
ax.set_xticks(x)
ax.set_xticklabels(pivot_data.index, rotation=45)
ax.legend()

# Add percentage labels on each section of the stacked bars
for i in range(len(pivot_data)):
    attended_count = pivot_data['count_of_appointments']['Attended'].iloc[i]
    dna_count = pivot_data['count_of_appointments']['DNA'].iloc[i]
    
    if attended_count > 0:
        # Attended segment percentage annotation
        ax.text(i, attended_count / 2, f'{pivot_data["percentage"]["Attended"].iloc[i]:.1f}%', ha='center', color='white')
    
    if dna_count > 0:
        # DNA segment percentage annotation on top of the 'DNA' segment
        ax.text(i, attended_count + dna_count + 2, f'{pivot_data["percentage"]["DNA"].iloc[i]:.1f}%', ha='center', color='black')

# Adjust layout for better visibility
plt.tight_layout()

# Show the plot
plt.show()
```

### Key Observations:

#### Face-to-Face Appointments Dominance: 
The majority of appointments are conducted face-to-face, with 94.2% attended and 5.8% DNA (Did Not Attend). This indicates a strong preference or requirement for in-person appointments.

#### High Attendance via Telephone: 
Telephone appointments show the highest attendance rate at 98%, with only 2% DNA. This suggests that telephone appointments may lead to higher attendance, possibly due to their convenience and accessibility.

#### Low Utilization of Home Visits and Video/Online Appointments: 
Home Visits and Video/Online appointments are minimal compared to Face-to-Face and Telephone modes. Both modes also show relatively high attendance, with 95.1% for Home Visits and 95.4% for Video/Online appointments, indicating that those who book these services are likely to attend.

### Recommendations:

Given the high attendance rate for telephone appointments, expanding telemedicine options, especially for routine consultations, could enhance patient access and reduce DNAs.

#### Promote Online Appointment Modes: 
Since Video/Online appointments have a relatively high attendance rate, promoting these as alternatives to face-to-face appointments may benefit patients who face barriers to in-person visits, such as mobility issues or distance.

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Assuming 'ar' is your DataFrame and it contains the following columns:
# 'appointment_mode', 'time_between_book_and_appointment', 'count_of_appointments'

# Step 1: Filter out entries with 'Unknown / Data Quality' in 'time_between_book_and_appointment' and 'Unknown' in 'appointment_mode'
filtered_data = ar[
    (ar['appointment_mode'] != 'Unknown') & 
    (ar['time_between_book_and_appointment'] != 'Unknown / Data Quality')
]

# Step 2: Calculate total counts for each 'time_between_book_and_appointment' and 'appointment_mode'
appointment_counts = filtered_data.groupby(['time_between_book_and_appointment', 'appointment_mode'])['count_of_appointments'].sum().reset_index()

# Step 3: Calculate total appointments per 'time_between_book_and_appointment'
total_per_time = appointment_counts.groupby('time_between_book_and_appointment')['count_of_appointments'].transform('sum')

# Step 4: Calculate the percentage of appointments for each 'appointment_mode' in each 'time_between_book_and_appointment'
appointment_counts['percentage'] = (appointment_counts['count_of_appointments'] / total_per_time) * 100

# Step 5: Define a consistent order for 'time_between_book_and_appointment' if necessary
# Here, we'll define a specific order to ensure the x-axis is logically ordered
time_order = [
    'Same Day', '1 Day', '2 to 7 Days', '8 to 14 Days',
    '15 to 21 Days', '22 to 28 Days', 'More than 28 Days'
]

# Step 6: Pivot data for the stacked bar plot
pivot_data = appointment_counts.pivot(index='time_between_book_and_appointment', columns='appointment_mode', values=['count_of_appointments', 'percentage']).fillna(0)
pivot_data = pivot_data.reindex(time_order)  # Reorder the index based on 'time_order'

# Step 7: Define colors for each appointment mode
appointment_colors = {
    'Face-to-Face': '#1f77b4',   # Blue
    'Telephone': '#ff7f0e',      # Orange
    'Home Visit': '#2ca02c',     # Green
    'Video/Online': '#d62728'    # Red
}

# Add a color for any additional modes if necessary
# For example, if there are modes not listed above, assign a default color
# Here, since 'Unknown' is filtered out, we don't need to handle it

# Step 8: Define x positions
x = np.arange(len(pivot_data.index))

# Step 9: Create the figure
fig, ax = plt.subplots(figsize=(16, 9))

# Initialize the bottom for stacking
bottom = np.zeros(len(pivot_data.index))

# Plot each appointment mode
for mode in appointment_colors.keys():
    counts = pivot_data[('count_of_appointments', mode)]
    ax.bar(x, counts, bottom=bottom, label=mode, color=appointment_colors[mode])
    bottom += counts  # Update the bottom for the next stack

# Set x-axis labels
ax.set_xticks(x)
ax.set_xticklabels(pivot_data.index, rotation=45, ha='right')

# Set title and labels
ax.set_title('Appointment Count and Percentage by Time Between Booking and Appointment Mode', fontsize=16)
ax.set_xlabel('Time Between Booking and Appointment', fontsize=14)
ax.set_ylabel('Count of Appointments', fontsize=14)

# Add a legend
ax.legend(title='Appointment Mode', fontsize=12, title_fontsize=12)

# Add percentage labels on each section of the stacked bars
for i in range(len(x)):
    cumulative = 0
    for mode in appointment_colors.keys():
        count = pivot_data[('count_of_appointments', mode)].iloc[i]
        percent = pivot_data[('percentage', mode)].iloc[i]
        if count > 0:
            ax.text(
                x[i],
                cumulative + count / 2,
                f'{percent:.1f}%',
                ha='center',
                va='center',
                color='white',
                fontsize=10,
                fontweight='bold'
            )
            cumulative += count

# Adjust layout for a clean presentation
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

# Define the list of relevant 'icb_ons_code' values
top_relevant_icb_codes = ["E54000050", "E54000054", "E54000057", "E54000008", "E54000027"]
bottom_relevant_icb_codes = ["E54000036", "E54000043", "E54000034", "E54000038", "E54000011"]

# Filter the data for the relevant 'icb_ons_code' and exclude 'Unknown' appointment modes
ar_filtered_top = ar[(ar['icb_ons_code'].isin(top_relevant_icb_codes)) & (ar['appointment_mode'] != 'Unknown')]
ar_filtered_bottom = ar[(ar['icb_ons_code'].isin(bottom_relevant_icb_codes)) & (ar['appointment_mode'] != 'Unknown')]

# Define a function to create the stacked bar charts with percentage annotations
def plot_stacked_bars(data, title, ax):
    # Group by 'icb_ons_code' and 'appointment_mode' and sum 'count_of_appointments'
    grouped_data = data.groupby(['icb_ons_code', 'appointment_mode'], as_index=False)['count_of_appointments'].sum()

    # Calculate percentage within each 'icb_ons_code'
    grouped_data['percentage'] = grouped_data['count_of_appointments'] / grouped_data.groupby('icb_ons_code')['count_of_appointments'].transform('sum') * 100

    # Pivot the data for counts and percentages
    pivot_counts = grouped_data.pivot(index='icb_ons_code', columns='appointment_mode', values='count_of_appointments').fillna(0)
    pivot_percentage = grouped_data.pivot(index='icb_ons_code', columns='appointment_mode', values='percentage').fillna(0)

    # Define a color palette for 'appointment_mode'
    appointment_modes = grouped_data['appointment_mode'].unique()
    colors = sns.color_palette("Set2", len(appointment_modes))
    color_dict = dict(zip(appointment_modes, colors))

    # Sort ICB codes by total appointments in descending order
    pivot_counts['total'] = pivot_counts.sum(axis=1)
    pivot_counts = pivot_counts.sort_values('total', ascending=False).drop('total', axis=1)
    pivot_percentage = pivot_percentage.loc[pivot_counts.index]
    
    x = np.arange(len(pivot_counts.index))  # label locations

    # Plot stacked bars
    bottom_counts = np.zeros(len(pivot_counts))
    for mode in appointment_modes:
        ax.bar(
            x,
            pivot_counts[mode],
            bottom=bottom_counts,
            label=mode if ax.get_legend() is None else "",  # Only add label once for legend
            color=color_dict.get(mode, '#333333')
        )
        bottom_counts += pivot_counts[mode]

    # Add title and labels
    ax.set_title(title, fontsize=14)
    ax.set_xlabel('ICB ONS Code')
    ax.set_ylabel('Count of Appointments')
    ax.set_xticks(x)
    ax.set_xticklabels(pivot_counts.index, rotation=45, ha='right')

    # Add percentage annotations
    for i, icb_code in enumerate(pivot_counts.index):
        cumulative_count = 0
        for mode in appointment_modes:
            count = pivot_counts.loc[icb_code, mode]
            percent = pivot_percentage.loc[icb_code, mode]
            if count > 0:
                y_position = cumulative_count + count / 2
                text_color = 'white' if percent > 10 else 'black'
                ax.text(
                    i,
                    y_position,
                    f'{percent:.1f}%',
                    ha='center',
                    va='center',
                    color=text_color,
                    fontsize=10,
                    fontweight='bold'
                )
                cumulative_count += count

# Create the subplots side by side
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(18, 8))

# Plot top 5 locations
plot_stacked_bars(ar_filtered_top, 'Top 5 Regions by Appointment Counts and Percentages', axes[0])

# Plot bottom 5 locations
plot_stacked_bars(ar_filtered_bottom, 'Bottom 5 Regions by Appointment Counts and Percentages', axes[1])

# Add legend to the first subplot only
axes[0].legend(title='Appointment Mode', bbox_to_anchor=(1.05, 1), loc='upper left')

# Adjust layout for better spacing
plt.tight_layout()

# Show the plot
plt.show()
```
